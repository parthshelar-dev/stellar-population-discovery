```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/gaia_cleaned.csv')

df.hist(bins=50, figsize=(15, 10), color='steelblue', edgecolor='black')
plt.suptitle('Feature Distributions', fontsize=16)
plt.tight_layout()
plt.savefig('../images/feature_distributions.png')
plt.show()
```

```python
plt.figure(figsize=(10, 8))
plt.scatter(df['bp_rp'], df['phot_g_mean_mag'], 
            alpha=0.1, s=1, c='steelblue')
plt.xlabel('BP-RP Color Index (Blue -> Red)', fontsize=12)
plt.ylabel('G Magnitude (Brightness)', fontsize=12)
plt.title('Hertzsprung-Russell Diagram', fontsize=16)
plt.gca().invert_yaxis()
plt.savefig('../images/hr_diagram.png')
plt.show()
```

```python
plt.style.use('default')
plt.figure(figsize=(10, 8))

scatter = plt.scatter(df['pmra'], df['pmdec'],
            alpha=0.2, s=1,
            c=df['teff_gspphot'],
            cmap='RdYlBu')

plt.colorbar(scatter, label='Surface Temperature (K)')
plt.xlabel('Proper Motion RA (mas/yr)', fontsize=12)
plt.ylabel('Proper Motion Dec (mas/yr)', fontsize=12)
plt.title('Stellar Proper Motion Colored by Temperature', fontsize=16)
plt.xlim(-50, 50)
plt.ylim(-50, 50)
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.savefig('../images/proper_motion.png', dpi=150, bbox_inches='tight')
plt.show()
```
