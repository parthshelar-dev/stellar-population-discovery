```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/gaia_raw.csv')

df = df.drop(columns=['SOURCE_ID', 'radial_velocity'])

df['teff_gspphot'] = df['teff_gspphot'].fillna(df['teff_gspphot'].median())

print("Null values after cleaning :\n", df.isnull().sum())
print("\nShape after cleaning :", df.shape)
print("\nRemaining columns :", df.columns.tolist())
```

```python
df.to_csv('../data/gaia_cleaned.csv', index=False)
print("Cleaned data saved successfully!")
```
